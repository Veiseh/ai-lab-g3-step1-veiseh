```python

# بررسی محیط اجرای پروژه تشخیص ژانر موسیقی
# در این بخش، نسخه پایتون، مسیر اجرا و سیستم‌عامل بررسی می‌شود.

import sys
import platform
from importlib import metadata

print("=== Environment Information ===")
print("Python version:", sys.version.split()[0])
print("Python executable:", sys.executable)
print("Operating system:", platform.platform())

# بررسی نسخه کتابخانه‌های موردنیاز پروژه
# این بخش نصب بودن کتابخانه‌ها و نسخه آن‌ها را نمایش می‌دهد.

print("\n=== Library Versions ===")

packages = [
    "numpy",
    "matplotlib",
    "librosa",
    "scikit-learn",
    "jupyter"
]

for package in packages:
    try:
        print(f"{package}: {metadata.version(package)}")
    except metadata.PackageNotFoundError:
        print(f"{package}: Not installed")

# آزمایش عملکرد کتابخانه‌های محاسباتی و رسم نمودار
# در این بخش، یک سیگنال سینوسی تولید و نمایش داده می‌شود.

print("\n=== Library Functionality Test ===")

try:
    import numpy as np
    import matplotlib.pyplot as plt

    x = np.linspace(0, 2 * np.pi, 100)
    y = np.sin(x)

    plt.plot(x, y)
    plt.title("Environment Test")
    plt.xlabel("Time")
    plt.ylabel("Amplitude")
    plt.grid(True)
    plt.show()

    print("NumPy and Matplotlib test completed successfully.")

except ImportError as error:
    print("Required library is missing:", error)

# جمع‌بندی بررسی محیط اجرا
# نتایج بالا برای اطمینان از آمادگی محیط پروژه ثبت می‌شوند.

print("\n=== Environment Check Completed ===")
```
